Coreccion del tipo de dato para: payer_type, breast_cancer_diagnosis_code y tipo_metastasis

```python
data['payer_type'] = data['payer_type'].astype('category')
data['breast_cancer_diagnosis_code'] = data['breast_cancer_diagnosis_code'].astype('category')
data['tipo_metastasis'] = data['tipo_metastasis'].astype('category')
data.info()
```

# 1. Smote

```python
print(data['tipo_metastasis'].value_counts())
data['tipo_metastasis'].value_counts().plot(kind='bar', grid=True, figsize=(5,3))
```

Se realiza un smote a la clase 'Equelitico' la cual cuenta con 1383 registros, el smote sera del 20 %.

```python
from imblearn.over_sampling import SMOTE, SMOTENC

X = data.drop("tipo_metastasis", axis = 1)
Y = data['tipo_metastasis']

smote = SMOTENC(k_neighbors=2,  sampling_strategy = {'Linfatico': 5876, 'VisceralSistemico': 2153, 'Esqueletico': 1659}, categorical_features=[0, 2]) # pyright: ignore[reportArgumentType]

X_smote, y_smote = smote.fit_resample(X,Y)

 # Creamos un dataframe con los resultados (X_smote, y_smote)
data  = pd.DataFrame(columns=X_smote.columns.values, data=X_smote)
data['tipo_metastasis']=y_smote
```

```python
data['tipo_metastasis'].value_counts().plot(kind='bar', grid=True)

print(data.shape)
print(print(data['tipo_metastasis'].value_counts()))
```
